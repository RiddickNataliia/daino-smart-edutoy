```python
# Install dependencies
!pip install roboflow ultralytics supervision matplotlib

import os
import cv2
import numpy as np
from ultralytics import YOLO
import supervision as sv
from roboflow import Roboflow
import matplotlib.pyplot as plt

# 1. Download dataset in YOLOv11 format
rf = Roboflow(api_key="pXbF9ik69UC691xELEwF")
project = rf.workspace("ai-concepts-ougid").project("daino_v2")
version = project.version(4)
dataset = version.download("yolov11")

# 2. Train YOLOv11n locally
model = YOLO("yolo11n.pt")  # Load pretrained YOLOv11n

# Train with validation (confusion matrix auto-generated)
model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    device=0  # Use GPU (0) or 'cpu'
)

# 3. Load best model and confusion matrix
best_model = YOLO(f"runs/detect/train/weights/best.pt")
confusion_matrix_path = "runs/detect/train/confusion_matrix.png"

# Display confusion matrix
if os.path.exists(confusion_matrix_path):
    img = cv2.imread(confusion_matrix_path)
    plt.figure(figsize=(12, 10))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title("Validation Confusion Matrix")
    plt.show()
else:
    print("Confusion matrix not found. Check training logs.")

# 4. Generate classification report
dataset_sv = sv.DetectionDataset.from_yolo(
    images_directory_path=f"{dataset.location}/valid/images",
    annotations_directory_path=f"{dataset.location}/valid/labels",
    data_yaml_path=f"{dataset.location}/data.yaml"
)

def callback(image: np.ndarray) -> sv.Detections:
    results = best_model(image)[0]
    return sv.Detections.from_ultralytics(results)

confusion_matrix = sv.ConfusionMatrix.benchmark(
    dataset=dataset_sv,
    callback=callback
)

print("\nClassification Report:")
print(f"Precision: {confusion_matrix.matrix.diagonal()/confusion_matrix.matrix.sum(axis=0)}")
print(f"Recall: {confusion_matrix.matrix.diagonal()/confusion_matrix.matrix.sum(axis=1)}")
print(f"F1 Score: {2 * (precision * recall) / (precision + recall)}")

# Plot confusion matrix with supervision
confusion_matrix.plot()
```
